Until now, Python chose and ran all the tools:

Now we will make the coach a little more autonomous.

**Autonomous Coach**: Gemini chooses the tools.

## 🧩 Challenge 9 — Give Gemini Access to Your Tools

We will use the same four functions from coach_toolbox_v4.py.

````python
from google.genai import types

gemini_tools = [
    ct.is_forward,
    ct.needs_shooting_practice,
    ct.needs_rest,
    ct.excellent_stamina
]
````

These are the tools.

We have not run them yet.

## 🤖 Build the Autonomous Coach

Create one function called:

````python
def autonomous_coach(question, player):

    context = get_coach_context(player)

    prompt = f'''
You are an AI Soccer Coach.

PLAYER DATA:
{context["player"]}

PERSISTENT MEMORY:
{context["memory"]}

SOCCER KNOWLEDGE:
{context["knowledge"]}

PLAYER QUESTION:
{question}

You have access to soccer decision tools.

Choose the tools that are useful for answering
the player's question.

You may use more than one tool if necessary.

Rules:
- Use the tools when they provide useful information.
- Do not invent player statistics.
- Do not invent tool results.
- Use only the available tools.
- Use the supplied memory when relevant.
- Use the supplied team data when relevant.
- Use the supplied soccer knowledge when relevant.
- Give practical and encouraging coaching advice.
'''

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            tools=gemini_tools,
            automatic_function_calling=(
                types.AutomaticFunctionCallingConfig(
                    maximum_remote_calls=5
                )
            )
        )
    )

    return response.text
````

### What Does maximum_remote_calls Mean?
`maximum_remote_calls=5`

sets the maximum number of automatic tool calls Gemini can make during one request.

It is a safety/control limit.

For example:

Gemini does not have to use all five calls.

It might use:

`1 call`

or

`2 calls`

or

`4 calls`

depending on what it determines is necessary.

The number 5 simply establishes the upper limit. 

If Gemini reaches the limit before it has enough information, the request may terminate without completing the intended tool-use sequence. Then the agent should stop trying tools and return a controlled response. It should not keep calling tools indefinitely.

## What Changed?

Compare the two approaches.

### Manual Tool Use

We first run:

`decisions = get_coach_decisions(selected_player)`

Then send the results to Gemini:

`context["decisions"]`

So Python decides which tools to run.

### Autonomous Coach

We give Gemini the functions:

`tools=gemini_tools`

Gemini can decide which tools it needs.

Python still executes the functions.

Gemini still receives the results and produces the final answer.
