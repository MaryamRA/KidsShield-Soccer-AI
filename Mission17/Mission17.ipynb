## 🧩 Challenge 9 — Give Gemini Access to Your Tools

We will use the same four functions from coach_toolbox_v4.py.

````python
# ============================================================
# SOCCER DECISION TOOLS
# ============================================================


def is_forward(player):
    """
    Determine whether a player plays as a forward.

    This function checks the player's position in the
    player dictionary.

    Args:
        player (dict):
            A dictionary containing the player's information.
            The dictionary should contain a "position" key.

    Returns:
        bool:
            True if the player's position is "Forward".
            False otherwise.

    Example:
        >>> player = {
        ...     "name": "Artin",
        ...     "position": "Forward"
        ... }
        >>> is_forward(player)
        True
    """

    return player["position"].lower() == "forward"


# ============================================================


def needs_shooting_practice(player):
    """
    Determine whether a player needs shooting practice.

    The function evaluates the player's shooting performance
    using the shooting-related information stored in the
    player dictionary.

    The player dictionary is expected to contain:
        - "shooting"
        - "total_goals"
        - "shots_on_target"

    Args:
        player (dict):
            A dictionary containing the player's soccer
            performance statistics.

    Returns:
        bool:
            True if the player's shooting performance indicates
            that additional shooting practice is recommended.
            False otherwise.

    Example:
        >>> player = {
        ...     "shooting": 88,
        ...     "total_goals": 6,
        ...     "shots_on_target": 5
        ... }
        >>> needs_shooting_practice(player)
        False
    """

    shooting = player["shooting"]
    total_goals = player["total_goals"]
    shots_on_target = player["shots_on_target"]

    # Shooting score below 70 indicates that
    # additional shooting practice is recommended.
    if shooting < 70:
        return True

    # Low goal production can also indicate
    # that additional shooting practice is useful.
    if total_goals < 5:
        return True

    # Few shots on target can indicate a need
    # for additional shooting practice.
    if shots_on_target < 3:
        return True

    return False


# ============================================================


def needs_rest(player):
    """
    Determine whether a player needs additional rest.

    The function evaluates the player's stamina level.

    The player dictionary is expected to contain:
        - "stamina"

    Args:
        player (dict):
            A dictionary containing the player's soccer
            performance statistics.

    Returns:
        bool:
            True if the player's stamina is below the
            recommended threshold and additional rest
            may be useful.
            False otherwise.

    Example:
        >>> player = {
        ...     "name": "Artin",
        ...     "stamina": 90
        ... }
        >>> needs_rest(player)
        False
    """

    stamina = player["stamina"]

    if stamina < 60:
        return True

    return False


# ============================================================


def excellent_stamina(player):
    """
    Determine whether a player's stamina is excellent.

    The function checks the player's stamina score against
    the threshold used by the coaching system.

    The player dictionary is expected to contain:
        - "stamina"

    Args:
        player (dict):
            A dictionary containing the player's soccer
            performance statistics.

    Returns:
        bool:
            True if the player's stamina is 80 or higher.
            False otherwise.

    Example:
        >>> player = {
        ...     "name": "Artin",
        ...     "stamina": 90
        ... }
        >>> excellent_stamina(player)
        True
    """

    stamina = player["stamina"]

    if stamina >= 80:
        return True

    return False
````

Here is the entire code for autonomous coach in one cell.

````python
def autonomous_coach(question, player):

    # Get the player's context
    context = get_coach_context(player)

    # ------------------------------------------------
    # Wrapper functions
    # ------------------------------------------------
    # These connect Gemini to coach_toolbox.
    # Each wrapper already knows which player
    # is being coached.

    def check_shooting():
        return ct.needs_shooting_practice(player)

    def check_rest():
        return ct.needs_rest(player)

    def check_stamina():
        return ct.excellent_stamina(player)

    def check_position():
        return ct.is_forward(player)

    # Give Gemini access to the real Python tools
    coach_tools = [
        check_shooting,
        check_rest,
        check_stamina,
        check_position
    ]

    # ------------------------------------------------
    # Give Gemini instructions
    # ------------------------------------------------

    prompt = f"""
You are an AI Soccer Coach.

PLAYER DATA:
{context["player"]}

PERSISTENT MEMORY:
{context["memory"]}

SOCCER KNOWLEDGE:
{context["knowledge"]}

You have access to real Python coaching tools.

IMPORTANT:
- Use a coach tool when it can answer the question.
- Never guess a result that a coach tool can determine.
- Use the tool result as the source of truth.
- Use memory and soccer knowledge to explain the result.
- Do not invent player information.

USER QUESTION:
{question}
"""

    # ------------------------------------------------
    # Gemini + real Python tools
    # ------------------------------------------------

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            tools=coach_tools
        )
    )

    return response.text
````

````python
# create a fime named: coach-memory_RAG_toolbox.py
# ============================================================
# ⚽ MISSION 17 — FINAL CHALLENGE
# Artin FC AI Soccer Agent
# Memory + Team Data + Soccer Knowledge + Coach Tools
# ============================================================

import os
import pandas as pd
import streamlit as st

from google import genai

import coach_toolbox_v4 as ct
import memory_toolbox as mt
import knowledge_toolbox as kt

from Artin_FC_data import Artin_FC_v3


# ============================================================
# PAGE SETTINGS
# ============================================================

st.set_page_config(
    page_title="Artin FC AI Soccer Agent",
    page_icon="⚽",
    layout="centered"
)


st.title("⚽ Artin FC AI Soccer Agent")

st.caption(
    "Mission 17 — Coach Toolbox + Manual & Autonomous AI"
)


st.write(
    """
Your AI Soccer Agent can now use four sources
of information:

🧠 Player Memory

⚽ Team Data

📚 Soccer Knowledge

🛠️ Coach Decision Tools
"""
)


# ============================================================
# LOAD MEMORY SAFELY
# ============================================================

def load_memory():

    if os.path.exists(mt.MEMORY_FILE_NAME):

        return mt.recall()

    else:

        return {}


# ============================================================
# LOAD PLAYER MEMORY
# ============================================================

memory = load_memory()


# ============================================================
# SECTION 1 — TEACH THE AGENT
# ============================================================

st.header("🧠 Teach My Agent")

st.write(
    "Add new information to your AI Soccer Agent's memory."
)


memory_key = st.text_input(
    "Memory Key",
    placeholder="Example: position"
)


memory_value = st.text_input(
    "Memory Value",
    placeholder="Example: Right Winger"
)


if st.button("💾 Remember"):

    if memory_key and memory_value:

        memory = mt.memorize(
            memory,
            memory_key,
            memory_value
        )

        mt.remember(
            memory,
            mt.MEMORY_FILE_NAME
        )

        st.success(
            "New information saved to memory!"
        )

    else:

        st.warning(
            "Please enter both a key and a value."
        )


# ============================================================
# SECTION 2 — SHOW MEMORY
# ============================================================

st.header("📖 My Soccer Memory")


memory = mt.recall()


if memory:

    st.json(memory)

else:

    st.info(
        "The agent does not have any memories yet."
    )


# ============================================================
# SECTION 3 — TEAM DATA
# ============================================================

st.header("⚽ Artin FC Team Data")


team_data = pd.DataFrame(
    Artin_FC_v3
)


st.dataframe(
    team_data,
    use_container_width=True
)


# ============================================================
# SECTION 4 — SELECT A PLAYER
# ============================================================

st.header("👤 Select a Player")


selected_player_name = st.selectbox(
    "Choose a player:",
    team_data["name"].tolist()
)


# Find selected player's row

player_data = team_data[
    team_data["name"] == selected_player_name
]


# Convert selected player to dictionary

selected_player = player_data.iloc[0].to_dict()


st.write(
    f"### {selected_player_name}'s Team Data"
)


st.dataframe(
    player_data,
    use_container_width=True
)


# ============================================================
# SECTION 5 — LOAD SOCCER KNOWLEDGE
# ============================================================

st.header("📚 Soccer Knowledge")


try:

    knowledge = kt.load_knowledge(
        "soccer_knowledge.txt"
    )

    st.success(
        "Soccer knowledge loaded successfully."
    )

except FileNotFoundError:

    knowledge = ""

    st.error(
        "soccer_knowledge.txt was not found."
    )


# ============================================================
# SECTION 6 — SEARCH SOCCER KNOWLEDGE
# ============================================================

st.subheader(
    "🔎 Search Soccer Knowledge"
)


keyword = st.text_input(
    "Enter a soccer topic:",
    value="winger"
)


if st.button(
    "📚 Search Knowledge"
):

    if knowledge:

        retrieved_results = kt.search_knowledge(
            knowledge,
            keyword
        )


        if retrieved_results:

            retrieved_info = "\n\n".join(
                retrieved_results
            )


            st.subheader(
                "📖 Retrieved Soccer Knowledge"
            )


            st.write(
                retrieved_info
            )


        else:

            st.warning(
                "No matching soccer knowledge was found."
            )

    else:

        st.warning(
            "The knowledge base is empty."
        )


# ============================================================
# SECTION 7 — COACH DECISION TOOLS
# ============================================================

st.header("🛠️ Coach Decision Tools")


def get_coach_decisions(player):

    decisions = {

        "is_forward":
            ct.is_forward(player),

        "needs_shooting_practice":
            ct.needs_shooting_practice(player),

        "needs_rest":
            ct.needs_rest(player),

        "excellent_stamina":
            ct.excellent_stamina(player)

    }

    return decisions


# Run the Python tools

decisions = get_coach_decisions(
    selected_player
)


st.write(
    "Python has evaluated the selected player:"
)


st.json(
    decisions
)


# ============================================================
# SECTION 8 — BUILD COACH CONTEXT
# ============================================================

def get_coach_context(player):

    memory = mt.recall()

    decisions = get_coach_decisions(
        player
    )

    if knowledge:

        retrieved_results = kt.search_knowledge(
            knowledge,
            keyword
        )

        if retrieved_results:

            retrieved_info = "\n\n".join(
                retrieved_results
            )

        else:

            retrieved_info = (
                "No matching soccer knowledge was found."
            )

    else:

        retrieved_info = (
            "No soccer knowledge is available."
        )


    context = {

        "player": player,

        "memory": memory,

        "knowledge": retrieved_info,

        "decisions": decisions

    }

    return context


# ============================================================
# SECTION 9 — MANUAL COACH
# ============================================================

st.header("🤖 Manual AI Coach")


st.write(
    """
In Manual Coach mode, Python runs the coach tools first.
Gemini receives the results and explains them.
"""
)


def ask_coach(question, player):

    context = get_coach_context(
        player
    )


    prompt = f"""
You are an AI Soccer Coach for Artin FC.

You have four sources of information.

============================================================
SOURCE 1 — PLAYER MEMORY
============================================================

{context["memory"]}


============================================================
SOURCE 2 — TEAM DATA
============================================================

{context["player"]}


============================================================
SOURCE 3 — SOCCER KNOWLEDGE
============================================================

{context["knowledge"]}


============================================================
SOURCE 4 — COACH TOOL DECISIONS
============================================================

{context["decisions"]}


============================================================
PLAYER QUESTION
============================================================

{question}


============================================================
INSTRUCTIONS
============================================================

Use the four sources to provide
personalized soccer coaching advice.

Respect the results of the Python tools.

Use the player's memory when relevant.

Use the player's team statistics when relevant.

Use the soccer knowledge when relevant.

Do not invent statistics.

Do not invent facts about the player.

If information is not available,
say that it is not available.

Give simple, practical advice that
a young soccer player can understand.
"""


    client = genai.Client()


    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )


    return response.text


# ============================================================
# SECTION 10 — AUTONOMOUS COACH
# ============================================================

st.header("🤖 Autonomous AI Coach")


st.write(
    """
In Autonomous Coach mode, Gemini will be given access
to the available Python tools and will decide which
tools are useful for answering the question.
"""
)


# ------------------------------------------------------------
# 🚧 PLACEHOLDER — ARTIN WILL BUILD THIS
# ------------------------------------------------------------

def autonomous_coach(question, player):

    """
    TODO — Build the autonomous coach.

    The autonomous coach should:

    1. Give Gemini access to the approved coach tools.
    2. Let Gemini decide which tools are useful.
    3. Allow Python to execute the selected tools.
    4. Return the tool results to Gemini.
    5. Let Gemini produce the final coaching advice.

    Hint:

    The tools should be provided through the Gemini
    generation configuration rather than being placed
    directly into the prompt as pre-computed decisions.
    """

    # --------------------------------------------------------
    # TODO: Artin builds the autonomous version here.
    # --------------------------------------------------------

    return (
        "🚧 Autonomous Coach placeholder.\n\n"
        "Artin's challenge: connect Gemini to the "
        "coach toolbox and allow Gemini to choose "
        "which tools it needs."
    )


# ============================================================
# SECTION 11 — ASK THE COACH
# ============================================================

st.header("💬 Ask My AI Coach")


user_question = st.text_input(
    "Ask your coach:",
    value="What should I improve?"
)


coach_mode = st.radio(
    "Choose Coach Mode:",
    [
        "Manual Coach",
        "Autonomous Coach"
    ]
)


if st.button(
    "⚽ Ask Coach",
    type="primary"
):

    if not user_question:

        st.warning(
            "Please enter a question."
        )

        st.stop()


    if "GEMINI_API_KEY" not in os.environ:

        st.error(
            "GEMINI_API_KEY was not found."
        )

        st.stop()


    # ========================================================
    # RUN SELECTED COACH
    # ========================================================

    with st.spinner(
        "Your AI Coach is thinking..."
    ):

        if coach_mode == "Manual Coach":

            answer = ask_coach(
                user_question,
                selected_player
            )

        else:

            answer = autonomous_coach(
                user_question,
                selected_player
            )


    # ========================================================
    # DISPLAY ANSWER
    # ========================================================

    st.subheader(
        "🤖 AI Coach Advice"
    )


    st.write(
        answer
    )


# ============================================================
# SECTION 12 — FORGET MEMORY
# ============================================================

st.header("🗑️ Forget Something")


st.write(
    "Remove a piece of information from the agent's memory."
)


memory = mt.recall()


if memory:

    memory_key_to_forget = st.selectbox(
        "Choose something to forget:",
        list(memory.keys())
    )


    if st.button(
        "🗑️ Forget"
    ):

        memory = mt.forget(
            memory,
            memory_key_to_forget
        )


        mt.remember(
            memory,
            mt.MEMORY_FILE_NAME
        )


        st.success(
            f"'{memory_key_to_forget}' "
            "has been removed from memory."
        )


        st.rerun()


else:

    st.info(
        "There is nothing to forget."
    )


# ============================================================
# SECTION 13 — SHOW THE FOUR SOURCES
# ============================================================

st.divider()

st.header(
    "🧠⚽📚🛠️ My AI Coach's Four Sources"
)


col1, col2, col3, col4 = st.columns(4)


with col1:

    st.subheader(
        "🧠 Memory"
    )

    st.write(
        "What the agent remembers about the player."
    )


with col2:

    st.subheader(
        "⚽ Team Data"
    )

    st.write(
        "Player statistics from Artin_FC_v3."
    )


with col3:

    st.subheader(
        "📚 Knowledge"
    )

    st.write(
        "Soccer information retrieved from documents."
    )


with col4:

    st.subheader(
        "🛠️ Tools"
    )

    st.write(
        "Python functions that evaluate the player."
    )


# ============================================================
# SECTION 14 — FINAL MISSION MESSAGE
# ============================================================

st.divider()


st.success(
    """
🎉 Mission 17 Complete!

Your AI Soccer Agent can now:

🧠 Remember information permanently

⚽ Read team and player data

📚 Retrieve soccer knowledge

🛠️ Use Python soccer decision tools

🤖 Use a Manual Coach

🚀 Prepare for an Autonomous Coach

🏆 Provide personalized coaching advice
"""
)
````
