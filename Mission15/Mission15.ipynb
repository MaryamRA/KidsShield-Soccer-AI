## 🟠 Challenge 3 — Persistent Memory

When we store information in Python, it normally lives only while the program is running. Python keeps this information in its memory. 

````python
player_data = {
    "player": "Artin",
    "position": "Right Winger",
    "speed": 8.5
}
````

If we stop the program, data disappears. This is called volatile data. 

How can we keep the data? We can save the information to a file on the computer. One common format is JSON. When we run the program again, we can read the file and get player's information back.

Simple example

````python
import json

player_data = {
    "player": "Artin",
    "position": "Right Winger",
    "speed": 8.5
}

with open("artin_data.json", "w") as file:
    json.dump(player_data, file, indent=2)
````


This creates:

`player_data.json`

containing:

````python

{
  "player": "Artin",
  "position": "Right Winger",
  "speed": 8.5
}
````

Python memory is temporary. A JSON file lets us save our data so we can use it again later.

````python
# what about when the value is a list? What about duplications?

# here is a complete version of memorize() function

def memorize(memory, key, value):

    # If the key does not exist, create it
    if key not in memory:
        memory[key] = value

    # If the key contains a list, add the value only if it is new
    elif isinstance(memory[key], list):
        if value not in memory[key]:
            memory[key].append(value)
        else:
            print("This information is already in memory.")
            return memory

    # If the key contains a single value, update it
    else:
        memory[key] = value

    print("New information memorized!")

    return memory
````

### Step 4. Remove Data from Json File - Forget()

Sometimes information can become incorrect or is no longer useful.

For example, suppose we decide that `"Passing"` should no longer be listed as one of Artin's strengths.

We can create a forget() function.

````python
def forget(memory, key, value):

    if key in memory:
        del memory[key]
        print("Information forgotten!")
    else:
        print("This information is not in memory.")

    return memory
````

````python
## test forget function
forget (soccer_memory, "a","xxxx")
````

````python
# here is a more advanced version of forget() function
def forget(memory, key, value=None):

    if key not in memory:
        print("This information is not in memory.")
        return memory

    if isinstance(memory[key], list):

        if value in memory[key]:
            memory[key].remove(value)
            print("Information forgotten!")
        else:
            print("This information is not in memory.")

    else:

        if value is None or memory[key] == value:
            del memory[key]
            print("Information forgotten!")
        else:
            print("This information is not in memory.")

    return memory
````

## Challenge 4. Building memory_toolbox.py

Next, we create a library called memory_toolbox.py and save it in the Mission15 folder. We will import this library in our AI Soccer Agent application and use it. 
memory_toolbox.py includes 4 main functions: 


**Tool 1**
`remember()`

Save the agent's memory.

**Tool 2**
`recall()`

Find something in memory.

**Tool 3**
`memorize()`

Add something to memory.


**Tool 4**
`forget()`

Remove something from the agent's memory.

````python
# ============================================================
# ⚽ MISSION 15
# memory_tools.py
#
# Memory tools for the AI Soccer Agent
# ============================================================

import json

MEMORY_FILE_NAME = "soccer_memory.json"

# ============================================================
# SAVE MEMORY
# ============================================================

def remember(memory, MEMORY_FILE_NAME):

    with open(MEMORY_FILE_NAME, "w") as file:
        json.dump(memory, file, indent=2)

    print("Soccer data saved!")
    return memory 

# ============================================================
# LOAD MEMORY
# ============================================================

def recall():

    with open(MEMORY_FILE_NAME, "r") as file:
        memory = json.load(file)

    print("Soccer data loaded!")

    return memory 

# ============================================================
# ADD DATA TO MEMORY
# ============================================================

def memorize(memory, key, value):

    # If the key does not exist, create it
    if key not in memory:
        memory[key] = value

    # If the key contains a list, add the value only if it is new
    elif isinstance(memory[key], list):
        if value not in memory[key]:
            memory[key].append(value)
        else:
            print("This information is already in memory.")
            return memory

    # If the key contains a single value, update it
    else:
        memory[key] = value

    print("New information memorized!")

    return memory


# ============================================================
# REMOVE DATA FROM MEMORY
# ============================================================

def forget(memory, key, value=None):

    if key not in memory:
        print("This information is not in memory.")
        return memory

    if isinstance(memory[key], list):

        if value in memory[key]:
            memory[key].remove(value)
            print("Information forgotten!")
        else:
            print("This information is not in memory.")

    else:

        if value is None or memory[key] == value:
            del memory[key]
            print("Information forgotten!")
        else:
            print("This information is not in memory.")

    return memory
````
