## 🧠 Phase 1 — What is an LLM?

### Traditional Programming Flow
$$\text{Notebook} \longrightarrow \text{Python} \longrightarrow \text{Rules} \longrightarrow \text{Output}$$

### LLM-Powered Flow
$$\text{Notebook} \longrightarrow \text{Prompt} \longrightarrow \text{OpenAI API} \longrightarrow \text{LLM} \longrightarrow \text{Response}$$

> **💡 Analogy:**
> - **Standard Python** is like writing every single instruction yourself.
> - **An LLM** is like hiring a very knowledgeable assistant who understands context and natural instructions.

---

## ⚽ Phase 2 — What is an API?

### Simple Analogy: The Restaurant
$$\text{Customer} \longrightarrow \text{Waiter (API)} \longrightarrow \text{Kitchen (LLM)} \longrightarrow \text{Meal (Response)}$$

- **Customer:** Your Python application.
- **Waiter (API):** The communication system that delivers your request and brings back the answer.
- **Kitchen (LLM):** The powerful remote server processing the request and preparing the output.

Our Python program never talks directly to the AI's inner brain—it sends structured requests through an **API**.

---

## 🔑 Phase 3 — Protecting Secrets

An **API Key** acts as your account password and access token. We **never** hard-code API keys directly into our codebase.

### ❌ Bad Practice (Hard-coded)
```python
api_key = "sk-proj-xxxxxxxxxxxxxxxxxxxxxxxx"
```

### ✅ Good Practice (Environment Variables)
```python
import os

api_key = os.getenv("OPENAI_API_KEY")
```

Always store sensitive parameters in a `.env` file or safe environment variable configuration!

## 🧠 Phase 7 — Prompt Engineering Basics

The clarity and structure of your input determine the quality of the generated advice.

### ❌ Vague Prompt
> *"Tell me about soccer."*

### ✅ Structured & Context-Rich Prompt
> *"You are an elite UEFA coach. The player is: Age: 16, Position: Forward, Goals: 12, Stamina: 62. Give exactly three recommendations."*

````python
structured_prompt = """
You are an elite UEFA coach.

The player profile is:
- Age: 16
- Position: Forward
- Goals: 12
- Stamina: 62

Give exactly three actionable recommendations for this player.
"""

response = client.responses.create(
    model="gpt-5",
    input=structured_prompt
)

print(response.output_text)
````

## 🚀 Final Challenge: Build the Interactive AI Assistant Web App

Let's bundle everything into a complete Streamlit web application: `app_assistant.py`.

**Requirements Checklist:**
- ✔ Text input box for coach query
- ✔ Interactive submit button
- ✔ AI response output container
- ✔ OpenAI API integration
- ✔ `coach_toolbox_v2` logic
- ✔ Streamlit UI interface

````python
%%writefile app_assistant.py
import streamlit as st
from openai import OpenAI
import coach_toolbox_v2 as tb

# Page setup
st.set_page_config(page_title="AI Soccer Assistant", page_icon="⚽")
st.title("⚽ AI Soccer Assistant")
st.caption("Ask questions about your team, tactics, or training plans!")

# Initialize OpenAI Client
client = OpenAI()

# Sample Roster Data
squad_data = [
    {"name": "Artin", "position": "LW", "stamina": 90, "speed": 92, "goals": [2, 1, 0, 3]},
    {"name": "Messi", "position": "CF", "stamina": 85, "speed": 89, "goals": [2, 3, 1, 2]},
    {"name": "Jahanbakhsh", "position": "RW", "stamina": 80, "speed": 85, "goals": [0, 1, 0, 1]}
]

# User Query Input
user_query = st.text_input("💬 Ask your Soccer Coach Assistant:", placeholder="e.g., Create today's training plan for Artin.")

if st.button("Ask Coach AI"):
    if user_query.strip():
        # Construct dynamic context prompt using squad data & toolbox functions
        squad_context = "\n".join([
            f"- {p['name']} ({p['position']}): Speed {p['speed']}, Stamina {p['stamina']}, Total Goals {tb.calculate_goals(p)}"
            for p in squad_data
        ])
        
        system_prompt = f"""
        You are an expert tactical soccer coach AI assistant.
        
        Current Squad Context:
        {squad_context}
        
        Answer the coach's request professionally and concisely.
        Request: {user_query}
        """
        
        with st.spinner("Analyzing squad data and formulating advice..."):
            response = client.responses.create(
                model="gpt-5",
                input=system_prompt
            )
            
        st.success("**AI Coach Response:**")
        st.write(response.output_text)
    else:
        st.warning("Please enter a question before submitting.")
````
