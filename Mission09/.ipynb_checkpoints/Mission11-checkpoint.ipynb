### 🏠 Step 2: Create Main App (`app.py`)

```python
%%writefile app.py
import streamlit as st

st.set_page_config(page_title="Artin FC Dashboard", page_icon="⚽", layout="wide")

st.sidebar.title("⚽ Coach Menu")
st.sidebar.info("Use the sidebar to navigate between pages.")

st.title("⚽ Artin FC Coaching Dashboard")
st.subheader("Welcome Coach Artin!")

st.markdown("""
Welcome to your professional soccer analytics suite.

### 📌 Features:
* **🏃 Player Analysis:** View metrics, performance attributes, and radar/bar charts.
* **📈 Team Statistics:** Explore roster-wide averages and player comparisons.
* **🤖 AI Coach:** Consult Gemini for tactical insights and roster recommendations.
""")
```

### 👤 Step 4: Player Analysis Page (`pages/1_Player_Analysis.py`)

```python
%%writefile pages/1_Player_Analysis.py
import streamlit as st
import pandas as pd
from Artin_FC_data import df

st.set_page_config(page_title="Player Analysis", page_icon="🏃", layout="wide")
st.sidebar.title("⚽ Coach Menu")

st.title("🏃 Player Analysis")
player_name = st.selectbox("Choose Player:", df["name"])
player_data = df[df["name"] == player_name].iloc[0]

col1, col2, col3 = st.columns(3)
col1.metric("⚡ Speed", player_data["speed"])
col2.metric("💪 Stamina", player_data["stamina"])
col3.metric("🎯 Shooting", player_data["shooting"])

st.divider()
st.subheader(f"Attributes for {player_name}")
attributes = pd.DataFrame({
    "Attribute": ["Speed", "Stamina", "Shooting", "Passing", "Defending"],
    "Rating": [player_data["speed"], player_data["stamina"], player_data["shooting"], player_data["passing"], player_data["defending"]]
}).set_index("Attribute")
st.bar_chart(attributes)
```

### 📈 Step 5: Team Statistics Page (`pages/2_Team_Statistics.py`)

```python
%%writefile pages/2_Team_Statistics.py
import streamlit as st
import pandas as pd
from Artin_FC_data import df

st.set_page_config(page_title="Team Statistics", page_icon="📈", layout="wide")
st.sidebar.title("⚽ Coach Menu")

st.title("📈 Team Statistics")
st.metric("Average Team Speed", f"{df['speed'].mean():.1f}")
st.subheader("Speed Comparison")
st.bar_chart(df[["name", "speed"]].set_index("name"))
```

### 🤖 Step 6: AI Coach Page (`pages/3_AI_Coach.py`)

```python
%%writefile pages/3_AI_Coach.py
import streamlit as st
import google.generativeai as genai
from Artin_FC_data import df

st.set_page_config(page_title="AI Assistant Coach", page_icon="🤖", layout="wide")
st.sidebar.title("⚽ Coach Menu")

st.title("🤖 AI Assistant Coach")
try:
    genai.configure(api_key=st.secrets["GEMINI_API_KEY"])
    model = genai.GenerativeModel("gemini-1.5-flash")
except Exception as e:
    st.error(f"API Key missing or invalid: {e}")
    st.stop()

question = st.text_input("Ask Coach Gemini:")
if st.button("Get Tactical Advice"):
    if question:
        context = f"Artin FC Squad Context:\n{df[['name', 'position', 'speed', 'stamina', 'total_goals']].to_string()}\n\n"
        response = model.generate_content(context + question)
        st.markdown("### 🤖 Advice")
        st.write(response.text)
```
