### 🤖 Step 3: AI Coach Page 
Create a python file in the pages folder named `pages/AI_Coach.py`

```python
# pages/AI_Coach.py
import streamlit as st
import google.generativeai as genai
from Artin_FC_data import df

st.set_page_config(page_title="AI Assistant Coach", page_icon="🤖", layout="wide")
st.sidebar.title("⚽ Coach Menu")

st.title("🤖 AI Assistant Coach")

genai.configure(api_key=st.secrets["GEMINI_API_KEY"])
model = genai.GenerativeModel("gemini-1.5-flash")

question = st.text_input("Ask AI Coach:")
        
if st.button("Get Tactical Advice"):
    if question:

        prompt = f"""
You are an expert professional soccer coach.

Below is information about the Artin FC squad.

{df[['name', 'position', 'speed', 'stamina', 'total_goals']].to_string(index=False)}

User's Question:
{question}

Using the squad information above, provide clear, practical, and personalized coaching advice.
"""

        response = model.generate_content(prompt)

        st.markdown("### 🤖 AI Tactical Advice")
        st.write(response.text)
```

#### Why we need to_string(index=false) in the prompt?
When we send information to Gemini, it does not receive a Pandas DataFrame. It only receives plain text.

The method to_string() converts the DataFrame into a clean text table that Gemini can easily read.

Without `to_string()`

```python
prompt = f"""
Here is my team:

{df}

Answer the coach's question.
"""
```

Python automatically converts the DataFrame into text, but the formatting is not always ideal. Large DataFrames may be shortened with ..., and the row numbers (indexes) are included.

With `to_string(index=False)`

```python
prompt = f"""
Here is my team:

{df.to_string(index=False)}

Answer the coach's question.
"""
```

Gemini receives: 
   - Cleaner and easier for humans to read.
   - Easier for the AI to understand.
   - More professional for real-world AI applications.
   - Less likely to include unnecessary information like DataFrame indexes.

## How to Run the Entire Web Application?

### How Streamlit Multi-Page Apps Work

A Streamlit application has:

- app.py = the main entrance
- pages/ = additional application screens

You only need to run ` streamli run app.py` to execute your entire application.

Streamlit automatically creates navigation. The connection is created by the folder name, not Python code
